```python
# ------------------------------------------------
# VEHICLE NUMBER PLATE VALIDATOR ( Version 2 )
# ------------------------------------------------

print("=" * 55)
print("VEHICLE NUMBER PLATE VALIDATOR ( Version 2 )".center(55))
print("=" * 55)
print()


numplate = input("Enter your Number Plate: ")
if numplate[:3].isalpha() == False :
 print()
 print("Invalid: first 3 characters must be letters")
elif numplate[:2].isupper() == False:
 print()
 print("Invalid: first 3 Characters must be Capitalized")
elif numplate[3] != "-":
 print()
 print("Invalid format: A hyphen is required after the letters")
elif not numplate[4:].isdigit():
  print()
  print("Invalid: last 3 characters must be digits")
elif len(numplate) != 7:
  print()
  print(f"Invalid plate: expected 7 characters, but got {len(numplate)} ( e.g :- ( {numplate} ))")
else:
  print()
  print(f"Valid plate: {numplate.upper()} (Format Accepted)")


print()
print("=" * 55)
print("SEE YOU IN THE NEXT REPOSITORIES..!".center(55))
print("=" * 55)
```
